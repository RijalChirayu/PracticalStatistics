```python
import pandas as pd 
from sklearn.linear_model import LinearRegression
from data_loader import get_dataset

lifestat = get_dataset(url = "https://raw.githubusercontent.com/ageron/data/refs/heads/main/lifesat/lifesat.csv", 
                       file_name="lifestat")
lifestat.head()
```

```python
X = lifestat[["GDP per capita (USD)"]].values 
y = lifestat[["Life satisfaction"]].values  

fig, ax = plt.subplots()
ax.scatter(x = X ,y  = y)
ax.set_xlabel("GDP per capita (USD)")
ax.set_ylabel("Life satisfaction")
ax.grid(visible=True)
plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)

X_new = [[1_377.6]]
model.predict(X_new)
```

```python
#using k-Nearest neigbours 
from sklearn.neighbors import KNeighborsRegressor

model = KNeighborsRegressor(n_neighbors= 3)
model.fit(X , y)

print(model.predict(X_new))
```
